3. Extraire la première ligne, la dernière colonne, puis la deuxième colonne entière.

```python
import numpy as np
```

```python
arr = np.array(range(0, 20))
arr_matrixe  = np.reshape(arr, (5, 4))
display(arr_matrixe)
```

```python
forme = arr_matrixe.shape
dim = arr_matrixe.ndim
types = arr_matrixe.dtype
size = arr_matrixe.size

print(f"sa forme : {forme}, son nombre de dimensions: {dim}, son type : {types} et sa taille totale : {size}")
```

```python
print(f" seul élément (ligne d'indice 3, colonne d'indice 2): {arr_matrixe[3, 2]}")
print(f" dernier élément avec des indices négatifs: {arr_matrixe[-1, -1]}")
```

```python
print(f"Extraire la première ligne : {arr_matrixe[0,:]}, la dernière colonne : {arr_matrixe[:,-1]}, puis la deuxième colonne entière : {arr_matrixe[:,1]} ")
```

```python
print(f"Extraire par slicing le bloc central (lignes d'indice 1 à 3, colonnes d'indice 1 à 2) : {arr_matrixe[1:3, 1:2]}")
```

```python
print(f"Extraire une ligne sur deux: {arr_matrixe[2, 0:arr_matrixe.shape[1]//2]},puis inverser l'ordre des lignes: {np.flip(arr_matrixe,)}")
```

```python
print(arr_matrixe[[0, 2, 4], :])
print(f"La différence entre le découpage et l'extraction de liste d'index : \nLe premier fonctionne pour des intervalles continus, rapide pour de grands intervalles et elle conserve l'ordre des resultats, ms l'extraction par listes des indices fonctionne pour des intervalles discretes, lente pour des grands intervalles, et l'ordre d'es resultats determiner par l'ordre de la liste")
```

```python
res = arr_matrixe[arr_matrixe % 2 == 0]
res[res > 8] = 0
res
```

```python
result = np.where(arr_matrixe % 5 == 0)
result
```

```python
slice_arr = arr_matrixe[0:4, 1:4]
slice_arr[0,0] = 100
print(f"le slice :\n {slice_arr}")
print(f"Original :\n {arr_matrixe} ")
print(f"\n Nous constatant que lors de la modification du slice la matrice original aussi change, car le slice prend une vue qui est une autre façon de visualisation des données dans laquel les donnes sont partages donc toute modification de la vue inpacte aussi l'original car c'est le meme pointeur dans la memoire, par contre une copie est un nouveau tableau entierement independant, si on change dans elle n'impacte pas le tableau original")
```

```python
import pandas as pd
```

```python
df = pd.read_csv("Titanic-Dataset.csv")
```

```python
df
```

```python
age_numpy = df['Age'].values
fare_numpy = df['Fare'].values
```

```python
nbr_age_nan = np.sum(np.isnan(age_numpy))
nbr_age_nan
```

La premier lorsqu'il fait la moyenne trouve des valeurs nan dans le resultats reste nan (sensible aux valeurs manquantes), ms pour la deuxieme qui ignore les Nan, il fait le calcul normale sur des valeurs non Nan et donne un resultat float, et si tout le tableau contient des valeurs Nan, il returne Nan

```python
print(f"La moyenne avec des valeurs manquantes : {np.mean(age_numpy)}")
```

```python
print(f"La moyenne avec des valeurs non manquantes :{np.nanmean(age_numpy):.2f}")
```

* On observe après comparaison de la moyenne et de la médiane de fare que la distribution des données est asymetrique a droite
* D'apres l'ecart type nous constatons que les valeurs sont largement dispersées autour de la moyenne.





```python
print(f"moyenne : {np.nanmean(age_numpy)}, médiane :  {np.nanmedian(age_numpy)}, écart-type :  {np.nanstd(age_numpy)}, minimum :  {np.nanmin(age_numpy)}, maximum :  {np.nanmax(age_numpy)} et percentiles 25 :  {np.nanquantile(age_numpy, 0.25)} et 75:  {np.nanquantile(age_numpy, 0.75)} de age.")
print(f"moyenne : {np.nanmean(fare_numpy)}, médiane :  {np.nanmedian(fare_numpy)}, écart-type :  {np.nanstd(fare_numpy)}, minimum :  {np.nanmin(fare_numpy)}, maximum :  {np.nanmax(fare_numpy)} et percentiles 25 :  {np.nanquantile(fare_numpy, 0.25)} et 75:  {np.nanquantile(fare_numpy, 0.75)} de fare.")
```

```python
Q1 = np.nanquantile(fare_numpy, 0.25)
Q3 = np.nanquantile(fare_numpy, 0.75)
IQR = Q3 - Q1
IQR
```

```python
borne_sup = Q3 + 1.5 * IQR
borne_sup
```

```python
outliers = np.sum(fare_numpy > borne_sup)
outliers
```

```python
fare_log1p = np.log1p(fare_numpy)
```

Apres avoir appliquer la transformation logarithmique log(x + 1), la distribution devient symetrique et la moyenne et la mediane preque égaux.

```python
print(f"moyenne : {np.nanmean(fare_log1p)}, médiane :  {np.nanmedian(fare_log1p)}, écart-type :  {np.nanstd(fare_log1p)}de fare.")
```

```python
enfant = age_numpy[age_numpy < 12]
adulte = age_numpy[age_numpy >= 12]
```

```python
nbr_enfants = np.sum(age_numpy < 12)
nbr_enfants
```

```python
np.correlate(age_numpy[~np.isnan(age_numpy)], fare_log1p[~np.isnan(fare_log1p)])
```

d'apres Pearson correlation coefficient, La diagonale est toujours 1 car c'est la correaltion avec la variable elle-meme, et les autres nous voyant que il ya une correaltion entre les deux variables

```python
np.corrcoef(np.nan_to_num(age_numpy, np.median(age_numpy)), np.nan_to_num(fare_log1p,  np.median(fare_log1p)))
```
